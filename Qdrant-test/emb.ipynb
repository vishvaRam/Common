```python
from fastembed import SparseTextEmbedding
```

```python

text = "OpenSearch and Qdrant are used for hybrid retrieval in RAG systems."

model = SparseTextEmbedding(model_name="Qdrant/bm25")
embedding = list(model.embed([text]))[0]

print("Input text:")
print(text)
```

```python

print("\nSparse embedding summary:")
print(f"Non-zero terms: {len(embedding.indices)}")
print("First 10 token ids and weights:")
for idx, val in list(zip(embedding.indices, embedding.values))[:10]:
    print(f"token_id={idx}, weight={float(val):.6f}")
```
